```python
import pandas as pd

train = pd.read_csv("/train.csv")
test = pd.read_csv("/test.csv")

train.head()
```

```python
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer

# Separate predictors and target
X = train.drop("Rings", axis=1)
y = train["Rings"]

# Tell Python which column is categorical
categorical = ["Sex"]

# Convert Sex into numbers
preprocessor = ColumnTransformer(
    transformers=[
        ("cat", OneHotEncoder(drop="first"), categorical)
    ],
    remainder="passthrough"
)

# Create model pipeline
from sklearn.pipeline import Pipeline

model = Pipeline([
    ("preprocessor", preprocessor),
    ("regression", LinearRegression())
])

# Fit model
model.fit(X, y)
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline

# Separate predictors and target
X = train.drop("Rings", axis=1)
y = train["Rings"]

# Categorical column
categorical = ["Sex"]

# Preprocessing
preprocessor = ColumnTransformer(
    transformers=[
        ("cat", OneHotEncoder(drop="first"), categorical)
    ],
    remainder="passthrough"
)

# Full model pipeline
model = Pipeline([
    ("preprocessor", preprocessor),
    ("regression", LinearRegression())
])

# Train model
model.fit(X, y)
```

```python
predictions = model.predict(test)

# Remove negative predictions
predictions = predictions.clip(0)
```

```python
submission = pd.DataFrame({
    "id": test["id"],
    "Rings": predictions
})

submission.to_csv("submission.csv", index=False)

submission.head()
```

```python
predictions.min()
```

Model 2 Random Forest

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, r2_score
```

```python
X = train.drop("Rings", axis=1)
y = train["Rings"]
```

```python
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline

categorical = ["Sex"]

preprocessor = ColumnTransformer(
    transformers=[
        ("cat", OneHotEncoder(drop="first"), categorical)
    ],
    remainder="passthrough"
)

rf_model = Pipeline([
    ("preprocessor", preprocessor),
    ("random_forest", RandomForestRegressor(
        n_estimators=200,
        random_state=42
    ))
])
```

```python
rf_model.fit(X, y)
```

```python
rf_predictions = rf_model.predict(test)

rf_predictions[:10]
```

```python
rf_predictions = rf_predictions.clip(0)

rf_predictions.min()
```

```python
rf_submission = pd.DataFrame({
    "id": test["id"],
    "Rings": rf_predictions
})

rf_submission.to_csv("random_forest_submission.csv", index=False)

rf_submission.head()
```

```python
import pandas as pd
import statsmodels.api as sm

# Load data
carseats = pd.read_csv("Carseats.csv")

# Select variables
X = carseats[['Price', 'Urban', 'US']]
y = carseats['Sales']

# Convert categorical variables to numbers
X = pd.get_dummies(X, drop_first=True)

# Make sure everything is numeric
X = X.astype(float)

# Add intercept
X = sm.add_constant(X)

# Fit regression
model = sm.OLS(y, X).fit()

# View results
model.summary()
```

```python
model.summary()
```

```python
# Smaller model using only Price and US

X2 = carseats[['Price', 'US']]
y = carseats['Sales']

# Convert US (Yes/No) into dummy variable
X2 = pd.get_dummies(X2, drop_first=True)

# Convert to numeric
X2 = X2.astype(float)

# Add intercept
X2 = sm.add_constant(X2)

# Fit smaller model
small_model = sm.OLS(y, X2).fit()

# Show results
small_model.summary()
```

```python
small_model.conf_int()
```

```python
influence = small_model.get_influence()

influence.summary_frame()
```

```python
# Smaller model: Sales predicted by Price and US only

X_small = carseats[['Price', 'US']]
y = carseats['Sales']

# Convert US Yes/No into numbers
X_small = pd.get_dummies(X_small, drop_first=True)

# Force numeric values
X_small = X_small.astype(float)

# Add intercept
X_small = sm.add_constant(X_small)

# Fit smaller model
small_model = sm.OLS(y, X_small).fit()

small_model.summary()
```
